```python
# ============================================================
# 11. ADD CONDITION-SPECIFIC RECOMMENDATIONS
# ============================================================

import pandas as pd

# ============================================================
# STEP 1: LOAD YOUR ACTUAL DATASET
# ============================================================

df = pd.read_csv("/content/smartagri_crop_yield_dataset.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)


# ============================================================
# STEP 2: CONDITION-SPECIFIC RECOMMENDATION FUNCTION
# ============================================================

def get_condition_recommendations(row):

    recommendations = []

    # --------------------------------------------------------
    # Soil pH condition
    # --------------------------------------------------------

    if row["Soil_pH"] < 6:

        recommendations.append(
            "Soil pH is low. Check soil correction requirements."
        )

    elif row["Soil_pH"] > 7.5:

        recommendations.append(
            "Soil pH is high. Monitor soil suitability."
        )

    else:

        recommendations.append(
            "Soil pH is within a suitable observed range."
        )


    # --------------------------------------------------------
    # Nitrogen condition
    # --------------------------------------------------------

    if row["Nitrogen"] < 50:

        recommendations.append(
            "Nitrogen level is low. Check nutrient requirements."
        )

    elif row["Nitrogen"] > 100:

        recommendations.append(
            "Nitrogen level is high. Avoid unnecessary nitrogen application."
        )

    else:

        recommendations.append(
            "Nitrogen level is within the observed dataset range."
        )


    # --------------------------------------------------------
    # Phosphorus condition
    # --------------------------------------------------------

    if row["Phosphorus"] < 30:

        recommendations.append(
            "Phosphorus level is low. Check phosphorus requirement."
        )


    # --------------------------------------------------------
    # Potassium condition
    # --------------------------------------------------------

    if row["Potassium"] < 30:

        recommendations.append(
            "Potassium level is low. Check potassium requirement."
        )


    # --------------------------------------------------------
    # Rainfall condition
    # --------------------------------------------------------

    if row["Rainfall"] < 500:

        recommendations.append(
            "Rainfall is low. Consider additional irrigation."
        )

    elif row["Rainfall"] > 1200:

        recommendations.append(
            "Rainfall is high. Monitor excess water and waterlogging."
        )

    else:

        recommendations.append(
            "Rainfall is within the observed range."
        )


    # --------------------------------------------------------
    # Temperature condition
    # --------------------------------------------------------

    if row["Temperature"] < 20:

        recommendations.append(
            "Temperature is low. Monitor crop growth."
        )

    elif row["Temperature"] > 35:

        recommendations.append(
            "Temperature is high. Monitor heat stress and water requirement."
        )

    else:

        recommendations.append(
            "Temperature is within the observed range."
        )


    # --------------------------------------------------------
    # Humidity condition
    # --------------------------------------------------------

    if row["Humidity"] > 80:

        recommendations.append(
            "Humidity is high. Monitor fungal disease risk."
        )

    elif row["Humidity"] < 40:

        recommendations.append(
            "Humidity is low. Monitor crop water requirement."
        )

    else:

        recommendations.append(
            "Humidity is within the observed range."
        )


    return recommendations


# ============================================================
# STEP 3: TEST DIFFERENT FARMING CONDITIONS
# ============================================================

print("\n================================================")
print("      CONDITION-SPECIFIC RECOMMENDATIONS")
print("================================================")


# Use first 10 real records from your dataset
test_cases = df.head(10)


for index, row in test_cases.iterrows():

    print("\n--------------------------------------------")

    print("Farm Case:", index + 1)

    print("Crop:", row["Crop"])

    print("Soil Type:", row["Soil_Type"])

    print("Soil pH:", row["Soil_pH"])

    print("Nitrogen:", row["Nitrogen"])

    print("Phosphorus:", row["Phosphorus"])

    print("Potassium:", row["Potassium"])

    print("Temperature:", row["Temperature"])

    print("Rainfall:", row["Rainfall"])

    print("Humidity:", row["Humidity"])


    print("\nRecommendations:")

    recommendations = get_condition_recommendations(row)

    for recommendation in recommendations:

        print("-", recommendation)


# ============================================================
# STEP 4: CREATE RECOMMENDATION TABLE
# ============================================================

result = []

for index, row in df.head(20).iterrows():

    recommendations = get_condition_recommendations(row)

    result.append({
        "Crop": row["Crop"],
        "Soil_pH": row["Soil_pH"],
        "Nitrogen": row["Nitrogen"],
        "Rainfall": row["Rainfall"],
        "Humidity": row["Humidity"],
        "Recommendations": " | ".join(recommendations)
    })


recommendation_df = pd.DataFrame(result)


# ============================================================
# STEP 5: DISPLAY TABLE
# ============================================================

print("\n================================================")
print("        RECOMMENDATION RESULT TABLE")
print("================================================")

print(
    recommendation_df.to_string(index=False)
)


# ============================================================
# STEP 6: FINAL SUMMARY
# ============================================================

print("\n================================================")
print("                 SUMMARY")
print("================================================")

print(
    "Total farm cases tested:",
    len(recommendation_df)
)

print(
    "Conditions checked:",
    "Soil pH, Nitrogen, Phosphorus, Potassium,",
    "Temperature, Rainfall, Humidity"
)

print(
    "Recommendation generation completed successfully!"
)
```
