```python
# ============================================================
# BUILD AN AGRICULTURAL KNOWLEDGE BASE
# ============================================================


# STEP 1: Import Libraries

import pandas as pd


# ============================================================
# STEP 2: Create Agricultural Knowledge Base
# ============================================================

knowledge_data = {

    "Crop": [
        "Rice",
        "Wheat",
        "Maize",
        "Cotton",
        "Tomato",
        "Groundnut",
        "Sugarcane",
        "Potato"
    ],

    "Soil": [
        "Clayey or loamy soil",
        "Loamy and well-drained soil",
        "Well-drained loamy soil",
        "Black soil or loamy soil",
        "Loamy soil rich in organic matter",
        "Sandy loam soil",
        "Deep loamy soil",
        "Sandy loam soil"
    ],

    "Temperature": [
        "20-35 C",
        "15-25 C",
        "18-27 C",
        "21-30 C",
        "20-30 C",
        "25-30 C",
        "20-35 C",
        "15-25 C"
    ],

    "Rainfall": [
        "1000-2000 mm",
        "450-650 mm",
        "500-800 mm",
        "700-1200 mm",
        "600-1200 mm",
        "500-700 mm",
        "1500-2500 mm",
        "500-700 mm"
    ],

    "Irrigation": [
        "Maintain adequate water during growth",
        "Provide irrigation at important growth stages",
        "Irrigate when soil moisture is low",
        "Provide irrigation during dry periods",
        "Regular irrigation is required",
        "Provide moderate irrigation",
        "Requires regular and sufficient irrigation",
        "Maintain uniform soil moisture"
    ],

    "Fertilizer": [
        "Nitrogen, phosphorus and potassium based on soil test",
        "Nitrogen and phosphorus based on soil requirement",
        "Nitrogen, phosphorus and potassium",
        "Nitrogen and potassium management",
        "Balanced NPK and organic manure",
        "Phosphorus, potassium and calcium",
        "Nitrogen, phosphorus and potassium",
        "Nitrogen, phosphorus and potassium"
    ],

    "Common_Pests": [
        "Stem borer and leaf folder",
        "Aphids and termites",
        "Fall armyworm and stem borer",
        "Bollworm and aphids",
        "Fruit borer and whitefly",
        "Leaf miner and aphids",
        "Early shoot borer and whitefly",
        "Potato tuber moth and aphids"
    ],

    "Disease_Management": [
        "Monitor for blast and bacterial diseases",
        "Monitor for rust and powdery mildew",
        "Monitor for leaf blight",
        "Monitor for wilt and leaf spot",
        "Monitor for leaf curl and early blight",
        "Monitor for leaf spot and rust",
        "Monitor for red rot and smut",
        "Monitor for late blight"
    ],

    "Harvest": [
        "Harvest when grains become mature and dry",
        "Harvest when grains become golden and dry",
        "Harvest when cobs are mature",
        "Harvest mature cotton bolls",
        "Harvest mature fruits",
        "Harvest when leaves turn yellow",
        "Harvest when crop reaches maturity",
        "Harvest when plants mature and leaves dry"
    ]
}


# ============================================================
# STEP 3: Convert Data into DataFrame
# ============================================================

knowledge_base = pd.DataFrame(knowledge_data)

print("Agricultural Knowledge Base")
print("==============================================")

print(knowledge_base)


# ============================================================
# STEP 4: Display Knowledge Base Information
# ============================================================

print("\nNumber of Crops:")
print(len(knowledge_base))

print("\nAvailable Crops:")

for crop in knowledge_base["Crop"]:
    print("-", crop)


# ============================================================
# STEP 5: Create Crop Information Function
# ============================================================

def get_crop_information(crop_name):

    crop_name = crop_name.lower()

    result = knowledge_base[
        knowledge_base["Crop"].str.lower() == crop_name
    ]

    if result.empty:

        return None

    return result.iloc[0]


# ============================================================
# STEP 6: Search for a Crop
# ============================================================

crop_name = input("\nEnter crop name: ")

crop_information = get_crop_information(crop_name)


# ============================================================
# STEP 7: Display Crop Information
# ============================================================

if crop_information is not None:

    print("\n==============================================")
    print("CROP INFORMATION")
    print("==============================================")

    print("Crop:", crop_information["Crop"])

    print("\nSoil Requirement:")
    print(crop_information["Soil"])

    print("\nTemperature:")
    print(crop_information["Temperature"])

    print("\nRainfall:")
    print(crop_information["Rainfall"])

    print("\nIrrigation:")
    print(crop_information["Irrigation"])

    print("\nFertilizer:")
    print(crop_information["Fertilizer"])

    print("\nCommon Pests:")
    print(crop_information["Common_Pests"])

    print("\nDisease Management:")
    print(crop_information["Disease_Management"])

    print("\nHarvest Information:")
    print(crop_information["Harvest"])

else:

    print("\nCrop not found in the knowledge base.")


# ============================================================
# STEP 8: Create Farmer Query Function
# ============================================================

def answer_farmer_query(crop_name, topic):

    crop_information = get_crop_information(crop_name)

    if crop_information is None:

        return "Sorry, crop information is not available."

    topic = topic.lower()

    if topic == "soil":

        return crop_information["Soil"]

    elif topic == "temperature":

        return crop_information["Temperature"]

    elif topic == "rainfall":

        return crop_information["Rainfall"]

    elif topic == "irrigation":

        return crop_information["Irrigation"]

    elif topic == "fertilizer":

        return crop_information["Fertilizer"]

    elif topic == "pest":

        return crop_information["Common_Pests"]

    elif topic == "disease":

        return crop_information["Disease_Management"]

    elif topic == "harvest":

        return crop_information["Harvest"]

    else:

        return "Topic not available."


# ============================================================
# STEP 9: Farmer Query
# ============================================================

print("\n==============================================")
print("FARMER QUERY SYSTEM")
print("==============================================")

crop = input("Enter crop name: ")

topic = input(
    "Enter topic "
    "(soil / temperature / rainfall / irrigation / fertilizer / pest / disease / harvest): "
)

answer = answer_farmer_query(crop, topic)

print("\nFarmer Query Answer:")
print(answer)


# ============================================================
# STEP 10: Display Knowledge Base Structure
# ============================================================

print("\n==============================================")
print("KNOWLEDGE BASE STRUCTURE")
print("==============================================")

print("""
Crop
 |
 +-- Soil Requirement
 |
 +-- Temperature
 |
 +-- Rainfall
 |
 +-- Irrigation
 |
 +-- Fertilizer
 |
 +-- Common Pests
 |
 +-- Disease Management
 |
 +-- Harvest Information
""")


# ============================================================
# STEP 11: Example Queries
# ============================================================

print("\n==============================================")
print("EXAMPLE FARMER QUESTIONS")
print("==============================================")

print("1. What soil is suitable for rice?")
print("2. What temperature is suitable for wheat?")
print("3. How much rainfall does maize require?")
print("4. How should tomato be irrigated?")
print("5. What fertilizer is suitable for groundnut?")
print("6. What are common pests in cotton?")
print("7. What diseases affect potato?")
print("8. When should sugarcane be harvested?")


# ============================================================
# STEP 12: Completion Message
# ============================================================

print("\n==============================================")
print("Agricultural Knowledge Base Created Successfully!")
print("==============================================")
```
