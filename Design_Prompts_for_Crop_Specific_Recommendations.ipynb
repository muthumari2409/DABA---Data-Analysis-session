```python
# ============================================================
# DESIGN PROMPTS FOR CROP-SPECIFIC RECOMMENDATIONS
# ============================================================


# STEP 1: Import Library

import pandas as pd


# ============================================================
# STEP 2: Create Crop Information
# ============================================================

crop_data = {

    "Rice": {
        "soil": "Clayey or loamy soil",
        "temperature": "20-35 C",
        "rainfall": "1000-2000 mm",
        "water": "Maintain adequate water during growth",
        "fertilizer": "Nitrogen, phosphorus and potassium based on soil test",
        "pests": "Stem borer and leaf folder"
    },

    "Wheat": {
        "soil": "Loamy and well-drained soil",
        "temperature": "15-25 C",
        "rainfall": "450-650 mm",
        "water": "Provide irrigation at important growth stages",
        "fertilizer": "Nitrogen and phosphorus based on soil requirement",
        "pests": "Aphids and termites"
    },

    "Maize": {
        "soil": "Well-drained loamy soil",
        "temperature": "18-27 C",
        "rainfall": "500-800 mm",
        "water": "Irrigate when soil moisture is low",
        "fertilizer": "Nitrogen, phosphorus and potassium",
        "pests": "Fall armyworm and stem borer"
    },

    "Cotton": {
        "soil": "Black soil or loamy soil",
        "temperature": "21-30 C",
        "rainfall": "700-1200 mm",
        "water": "Provide irrigation during dry periods",
        "fertilizer": "Nitrogen and potassium management",
        "pests": "Bollworm and aphids"
    },

    "Tomato": {
        "soil": "Loamy soil rich in organic matter",
        "temperature": "20-30 C",
        "rainfall": "600-1200 mm",
        "water": "Regular irrigation is required",
        "fertilizer": "Balanced NPK and organic manure",
        "pests": "Fruit borer and whitefly"
    },

    "Groundnut": {
        "soil": "Sandy loam soil",
        "temperature": "25-30 C",
        "rainfall": "500-700 mm",
        "water": "Provide moderate irrigation",
        "fertilizer": "Phosphorus, potassium and calcium",
        "pests": "Leaf miner and aphids"
    }
}


# ============================================================
# STEP 3: Create Crop Recommendation Function
# ============================================================

def generate_crop_prompt(
    crop,
    nitrogen,
    phosphorus,
    potassium,
    temperature,
    rainfall,
    humidity,
    predicted_yield
):

    if crop not in crop_data:

        return "Crop information is not available."


    information = crop_data[crop]


    # Determine yield status

    if predicted_yield >= 4.5:

        yield_status = "High"

    elif predicted_yield >= 3.5:

        yield_status = "Medium"

    else:

        yield_status = "Low"


    # Determine nitrogen condition

    if nitrogen < 70:

        nitrogen_status = "Low"

    elif nitrogen <= 90:

        nitrogen_status = "Normal"

    else:

        nitrogen_status = "High"


    # Determine rainfall condition

    if rainfall < 700:

        rainfall_status = "Low"

    elif rainfall <= 1000:

        rainfall_status = "Moderate"

    else:

        rainfall_status = "High"


    # ========================================================
    # Create LLM Prompt
    # ========================================================

    prompt = f"""
You are an agricultural advisory assistant.

Your task is to provide simple and practical crop management
recommendations for farmers.

CROP INFORMATION
-------------------------
Crop: {crop}

Suitable Soil:
{information["soil"]}

Suitable Temperature:
{information["temperature"]}

Required Rainfall:
{information["rainfall"]}

Recommended Water Management:
{information["water"]}

Recommended Fertilizer:
{information["fertilizer"]}

Common Pests:
{information["pests"]}


CURRENT FARM CONDITIONS
-------------------------
Nitrogen: {nitrogen}
Nitrogen Status: {nitrogen_status}

Phosphorus: {phosphorus}

Potassium: {potassium}

Temperature: {temperature} C

Rainfall: {rainfall} mm
Rainfall Status: {rainfall_status}

Humidity: {humidity} %

Predicted Crop Yield: {predicted_yield}

Yield Status: {yield_status}


TASK
-------------------------
Based on the crop information and current farm conditions:

1. Explain the current crop condition.
2. Give soil management advice.
3. Give irrigation advice.
4. Give fertilizer advice.
5. Give pest monitoring advice.
6. Give yield improvement suggestions.
7. Provide immediate actions for the farmer.


IMPORTANT
-------------------------
- Use simple language.
- Give practical recommendations.
- Do not make unsupported claims.
- Consider the provided crop information.
- Mention when a soil test or agricultural expert is needed.
- Keep the answer easy for a farmer to understand.
"""

    return prompt


# ============================================================
# STEP 4: Enter Farm Information
# ============================================================

crop = input("Enter crop name (Rice/Wheat/Maize/Cotton/Tomato/Groundnut): ")

nitrogen = float(input("Enter Nitrogen value: "))

phosphorus = float(input("Enter Phosphorus value: "))

potassium = float(input("Enter Potassium value: "))

temperature = float(input("Enter Temperature in C: "))

rainfall = float(input("Enter Rainfall in mm: "))

humidity = float(input("Enter Humidity percentage: "))

predicted_yield = float(input("Enter Predicted Yield: "))


# ============================================================
# STEP 5: Generate Crop-Specific Prompt
# ============================================================

prompt = generate_crop_prompt(
    crop,
    nitrogen,
    phosphorus,
    potassium,
    temperature,
    rainfall,
    humidity,
    predicted_yield
)


# ============================================================
# STEP 6: Display Prompt
# ============================================================

print("\n")
print("============================================================")
print("CROP-SPECIFIC LLM PROMPT")
print("============================================================")

print(prompt)


# ============================================================
# STEP 7: Create Multiple Prompt Templates
# ============================================================

prompt_templates = {

    "Basic Recommendation":
    """
Provide a simple recommendation for the farmer based on
crop, soil, climate and predicted yield.
""",

    "Irrigation Recommendation":
    """
Analyze the rainfall, temperature and soil conditions.
Provide suitable irrigation advice for the crop.
""",

    "Fertilizer Recommendation":
    """
Analyze the soil nutrient values and provide fertilizer
management advice for the selected crop.
""",

    "Pest Monitoring":
    """
Identify important pest monitoring points for the selected
crop and provide safe management guidance.
""",

    "Yield Improvement":
    """
Analyze the predicted yield and farm conditions.
Provide practical steps to improve crop yield.
"""
}


# ============================================================
# STEP 8: Display Prompt Templates
# ============================================================

print("\n")
print("============================================================")
print("PROMPT TEMPLATES")
print("============================================================")

for name, template in prompt_templates.items():

    print("\n" + name)
    print("----------------------------")
    print(template)


# ============================================================
# STEP 9: Create Prompt Dataset
# ============================================================

prompt_dataset = pd.DataFrame({

    "Prompt_Type": [
        "Basic Recommendation",
        "Irrigation Recommendation",
        "Fertilizer Recommendation",
        "Pest Monitoring",
        "Yield Improvement"
    ],

    "Purpose": [
        "General crop management advice",
        "Water and irrigation management",
        "Soil nutrient and fertilizer management",
        "Pest monitoring and management",
        "Improve expected crop yield"
    ]
})


print("\n")
print("============================================================")
print("PROMPT DESIGN TABLE")
print("============================================================")

print(prompt_dataset)


# ============================================================
# STEP 10: Final Workflow
# ============================================================

print("\n")
print("============================================================")
print("CROP-SPECIFIC RECOMMENDATION WORKFLOW")
print("============================================================")

print("""
Farmer / Farm Data
        ↓
Crop Information
        ↓
Soil Information
        ↓
Climate Information
        ↓
Predicted Yield
        ↓
Prompt Generation
        ↓
LLM Agricultural Assistant
        ↓
Crop-Specific Recommendation
        ↓
Farmer Action
""")


# ============================================================
# STEP 11: Completion Message
# ============================================================

print("============================================================")
print("Crop-Specific Prompt Design Completed Successfully!")
print("============================================================")
```
