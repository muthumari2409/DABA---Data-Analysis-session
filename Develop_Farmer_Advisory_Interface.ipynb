```python
# ============================================================
# 12. DEVELOP FARMER ADVISORY INTERFACE
# ============================================================

import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output


# ============================================================
# STEP 1: LOAD YOUR ACTUAL DATASET
# ============================================================

df = pd.read_csv("/content/smartagri_crop_yield_dataset.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)


# ============================================================
# STEP 2: GET AVAILABLE CROPS
# ============================================================

crops = sorted(df["Crop"].unique().tolist())

print("\nAvailable Crops:")
for crop in crops:
    print("-", crop)


# ============================================================
# STEP 3: CREATE FARMER INTERFACE
# ============================================================

crop_dropdown = widgets.Dropdown(
    options=crops,
    description="Crop:",
    style={"description_width": "initial"}
)

question_dropdown = widgets.Dropdown(
    options=[
        "Soil information",
        "Rainfall information",
        "Temperature information",
        "Nitrogen information",
        "Phosphorus information",
        "Potassium information",
        "Humidity information",
        "Yield information",
        "General advisory"
    ],
    description="Question:",
    style={"description_width": "initial"}
)

button = widgets.Button(
    description="Get Farmer Advisory",
    button_style="success"
)

output = widgets.Output()


# ============================================================
# STEP 4: FARMER ADVISORY FUNCTION
# ============================================================

def generate_advisory(crop, question):

    crop_data = df[df["Crop"] == crop]

    if len(crop_data) == 0:

        return "Crop information is not available."


    # Average values for selected crop
    soil_type = crop_data["Soil_Type"].mode()[0]

    soil_ph = crop_data["Soil_pH"].mean()

    nitrogen = crop_data["Nitrogen"].mean()

    phosphorus = crop_data["Phosphorus"].mean()

    potassium = crop_data["Potassium"].mean()

    temperature = crop_data["Temperature"].mean()

    rainfall = crop_data["Rainfall"].mean()

    humidity = crop_data["Humidity"].mean()

    yield_value = crop_data["Yield"].mean()


    # --------------------------------------------------------
    # Soil
    # --------------------------------------------------------

    if question == "Soil information":

        return f"""
Crop: {crop}

Soil Type: {soil_type}

Average Soil pH: {soil_ph:.2f}

Advice:
Monitor the soil condition and maintain suitable soil management.
"""


    # --------------------------------------------------------
    # Rainfall
    # --------------------------------------------------------

    elif question == "Rainfall information":

        return f"""
Crop: {crop}

Average Rainfall: {rainfall:.2f}

Advice:
Monitor rainfall and provide irrigation when required.
"""


    # --------------------------------------------------------
    # Temperature
    # --------------------------------------------------------

    elif question == "Temperature information":

        return f"""
Crop: {crop}

Average Temperature: {temperature:.2f}

Advice:
Monitor temperature conditions during crop growth.
"""


    # --------------------------------------------------------
    # Nitrogen
    # --------------------------------------------------------

    elif question == "Nitrogen information":

        return f"""
Crop: {crop}

Average Nitrogen: {nitrogen:.2f}

Advice:
Check nitrogen availability and manage nutrients according to crop requirements.
"""


    # --------------------------------------------------------
    # Phosphorus
    # --------------------------------------------------------

    elif question == "Phosphorus information":

        return f"""
Crop: {crop}

Average Phosphorus: {phosphorus:.2f}

Advice:
Monitor phosphorus availability for crop growth.
"""


    # --------------------------------------------------------
    # Potassium
    # --------------------------------------------------------

    elif question == "Potassium information":

        return f"""
Crop: {crop}

Average Potassium: {potassium:.2f}

Advice:
Monitor potassium availability and maintain proper nutrient management.
"""


    # --------------------------------------------------------
    # Humidity
    # --------------------------------------------------------

    elif question == "Humidity information":

        return f"""
Crop: {crop}

Average Humidity: {humidity:.2f}

Advice:
Monitor humidity and watch for possible disease risk when humidity is high.
"""


    # --------------------------------------------------------
    # Yield
    # --------------------------------------------------------

    elif question == "Yield information":

        return f"""
Crop: {crop}

Average Recorded Yield: {yield_value:.2f}

Advice:
Continue monitoring soil, water and nutrient conditions to maintain crop yield.
"""


    # --------------------------------------------------------
    # General Advisory
    # --------------------------------------------------------

    else:

        advice = []

        if rainfall < 500:

            advice.append(
                "Rainfall is relatively low. Consider irrigation."
            )

        if humidity > 80:

            advice.append(
                "Humidity is high. Monitor disease risk."
            )

        if nitrogen < 50:

            advice.append(
                "Nitrogen is relatively low. Check nutrient requirements."
            )

        if soil_ph < 6:

            advice.append(
                "Soil pH is relatively low. Check soil correction requirements."
            )

        if len(advice) == 0:

            advice.append(
                "No major condition warning from the selected dataset values."
            )

        return (
            f"Crop: {crop}\n\n"
            f"Average Yield: {yield_value:.2f}\n\n"
            "Farmer Recommendations:\n- "
            + "\n- ".join(advice)
        )


# ============================================================
# STEP 5: BUTTON ACTION
# ============================================================

def show_advisory(button_click):

    with output:

        clear_output()

        crop = crop_dropdown.value

        question = question_dropdown.value

        print("==============================================")
        print("          SMART FARMER ADVISORY")
        print("==============================================")

        print("\nSelected Crop:", crop)

        print("Selected Query:", question)

        print("\n----------------------------------------------")
        print("                ADVISORY")
        print("----------------------------------------------")

        print(
            generate_advisory(
                crop,
                question
            )
        )

        print("\n==============================================")
        print("Advisory generated using your dataset.")
        print("==============================================")


button.on_click(show_advisory)


# ============================================================
# STEP 6: DISPLAY INTERFACE
# ============================================================

display(crop_dropdown)

display(question_dropdown)

display(button)

display(output)


# ============================================================
# STEP 7: SHOW ONE AUTOMATIC EXAMPLE
# ============================================================

crop_dropdown.value = crops[0]

question_dropdown.value = "General advisory"

show_advisory(None)
```
