```python
# ============================================================
# FARMER ADVISORY ASSISTANT - SIMPLE VERSION
# ============================================================

import pandas as pd


# STEP 1: Agricultural Knowledge Base

data = {
    "Crop": ["Rice", "Wheat", "Maize", "Cotton", "Tomato", "Groundnut"],

    "Soil": [
        "Clayey or loamy soil",
        "Loamy and well-drained soil",
        "Well-drained loamy soil",
        "Black soil or loamy soil",
        "Loamy soil rich in organic matter",
        "Sandy loam soil"
    ],

    "Rainfall": [
        "1000-2000 mm",
        "450-650 mm",
        "500-800 mm",
        "700-1200 mm",
        "600-1200 mm",
        "500-700 mm"
    ],

    "Irrigation": [
        "Maintain adequate water during growth",
        "Provide irrigation at important growth stages",
        "Irrigate when soil moisture is low",
        "Provide irrigation during dry periods",
        "Regular irrigation is required",
        "Provide moderate irrigation"
    ],

    "Fertilizer": [
        "Nitrogen, phosphorus and potassium based on soil test",
        "Nitrogen and phosphorus based on soil requirement",
        "Nitrogen, phosphorus and potassium",
        "Nitrogen and potassium management",
        "Balanced NPK and organic manure",
        "Phosphorus, potassium and calcium"
    ],

    "Pests": [
        "Stem borer and leaf folder",
        "Aphids and termites",
        "Fall armyworm and stem borer",
        "Bollworm and aphids",
        "Fruit borer and whitefly",
        "Leaf miner and aphids"
    ]
}


knowledge_base = pd.DataFrame(data)


# STEP 2: Farmer Question Function

def farmer_advisory(question):

    question = question.lower()

    # Find crop

    crop = None

    for name in knowledge_base["Crop"]:

        if name.lower() in question:

            crop = name
            break


    if crop is None:

        return "Crop name was not found."


    # Get crop information

    info = knowledge_base[
        knowledge_base["Crop"] == crop
    ].iloc[0]


    # Find topic

    if "soil" in question:

        answer = info["Soil"]

        topic = "Soil"

    elif "rainfall" in question or "rain" in question:

        answer = info["Rainfall"]

        topic = "Rainfall"

    elif "water" in question or "irrigation" in question:

        answer = info["Irrigation"]

        topic = "Irrigation"

    elif "fertilizer" in question or "fertiliser" in question:

        answer = info["Fertilizer"]

        topic = "Fertilizer"

    elif "pest" in question or "insect" in question:

        answer = info["Pests"]

        topic = "Pests"

    else:

        answer = "Please ask about soil, rainfall, irrigation, fertilizer or pests."

        topic = "General"


    # Final answer

    response = f"""
Crop: {crop}

Topic: {topic}

Farmer Advice:
{answer}

Please monitor the crop regularly and use soil-test
results when applying fertilizers.
"""

    return response


# STEP 3: Test Questions Automatically

questions = [
    "What soil is suitable for rice?",
    "How much rainfall does wheat need?",
    "What fertilizer is suitable for maize?",
    "What pests affect tomato?",
    "How should cotton be irrigated?",
    "What soil is suitable for groundnut?"
]


# STEP 4: Display Answers

print("================================================")
print("       FARMER ADVISORY ASSISTANT")
print("================================================")


for question in questions:

    print("\nFarmer Question:")
    print(question)

    print("\nAssistant Answer:")

    print(farmer_advisory(question))

    print("-----------------------------------------------")


print("\nFarmer Advisory Assistant Completed Successfully!")
```
