```python
# ============================================================
# 9. IDENTIFY UNSUITABLE OR UNSUPPORTED ADVICE
# ============================================================

import pandas as pd

# ============================================================
# STEP 1: LOAD YOUR ACTUAL DATASET
# ============================================================

df = pd.read_csv("/content/smartagri_crop_yield_dataset.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)


# ============================================================
# STEP 2: GET SUPPORTED CROPS
# ============================================================

supported_crops = df["Crop"].unique().tolist()

print("\nSupported Crops:")
for crop in supported_crops:
    print("-", crop)


# ============================================================
# STEP 3: ADVICE VALIDATION FUNCTION
# ============================================================

def validate_advice(crop, advice):

    # Check whether crop exists
    crop_exists = False

    for available_crop in supported_crops:

        if available_crop.lower() == crop.lower():
            crop_exists = True
            break

    # Unsupported crop
    if crop_exists == False:

        return "UNSUPPORTED - Crop not available in dataset"


    # Agricultural topics available in dataset
    supported_topics = [
        "soil",
        "ph",
        "nitrogen",
        "phosphorus",
        "potassium",
        "temperature",
        "rainfall",
        "rain",
        "humidity",
        "water",
        "irrigation",
        "fertilizer",
        "nutrient",
        "pest",
        "disease"
    ]

    advice_lower = advice.lower()

    topic_found = False

    for topic in supported_topics:

        if topic in advice_lower:
            topic_found = True
            break


    if topic_found:

        return "SUPPORTED"

    else:

        return "UNSUPPORTED - Topic is not available in dataset"


# ============================================================
# STEP 4: TEST FARMER ADVICE
# ============================================================

test_cases = [

    ("Rice", "Check soil pH"),
    ("Rice", "Improve nitrogen management"),

    ("Wheat", "Monitor rainfall"),
    ("Wheat", "Check irrigation requirement"),

    ("Maize", "Check potassium level"),
    ("Maize", "Monitor temperature"),

    ("Cotton", "Monitor humidity"),
    ("Cotton", "Check soil condition"),

    ("Tomato", "Check phosphorus level"),
    ("Tomato", "Monitor rainfall"),

    ("Groundnut", "Check nitrogen requirement"),
    ("Groundnut", "Monitor soil pH"),

    # Unsupported examples
    ("Banana", "Apply fertilizer"),
    ("Apple", "Check soil condition"),
    ("Rice", "Use satellite rocket technology")
]


# ============================================================
# STEP 5: VALIDATE EACH ADVICE
# ============================================================

results = []

for crop, advice in test_cases:

    status = validate_advice(
        crop,
        advice
    )

    results.append({
        "Crop": crop,
        "Advice": advice,
        "Status": status
    })


result_df = pd.DataFrame(results)


# ============================================================
# STEP 6: DISPLAY RESULTS
# ============================================================

print("\n================================================")
print("          ADVICE VALIDATION RESULTS")
print("================================================")

print(
    result_df.to_string(index=False)
)


# ============================================================
# STEP 7: COUNT SUPPORTED AND UNSUPPORTED
# ============================================================

supported_count = 0
unsupported_count = 0

for status in result_df["Status"]:

    if status == "SUPPORTED":
        supported_count += 1

    else:
        unsupported_count += 1


total = len(result_df)


# ============================================================
# STEP 8: CALCULATE PERCENTAGE
# ============================================================

supported_percentage = (
    supported_count / total
) * 100

unsupported_percentage = (
    unsupported_count / total
) * 100


# ============================================================
# STEP 9: FINAL SUMMARY
# ============================================================

print("\n================================================")
print("              VALIDATION SUMMARY")
print("================================================")

print("Total advice tested:", total)

print(
    "Supported advice:",
    supported_count
)

print(
    "Unsupported advice:",
    unsupported_count
)

print(
    "Supported percentage:",
    round(supported_percentage, 2),
    "%"
)

print(
    "Unsupported percentage:",
    round(unsupported_percentage, 2),
    "%"
)


# ============================================================
# STEP 10: SHOW UNSUPPORTED ADVICE
# ============================================================

print("\n================================================")
print("          UNSUPPORTED ADVICE")
print("================================================")

unsupported_rows = result_df[
    result_df["Status"] != "SUPPORTED"
]

if len(unsupported_rows) > 0:

    print(
        unsupported_rows.to_string(index=False)
    )

else:

    print("No unsupported advice found.")


print("\nAdvice validation completed successfully!")
```
