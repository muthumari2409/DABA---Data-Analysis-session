```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer(
    stop_words="english"
)

document_vectors = vectorizer.fit_transform(documents)

print("Knowledge base indexed successfully!")
```

```python
def retrieve_context(question, top_k=2):

    question_vector = vectorizer.transform([question])

    similarity_scores = cosine_similarity(
        question_vector,
        document_vectors
    ).flatten()

    top_indices = similarity_scores.argsort()[::-1][:top_k]

    retrieved_documents = []

    for index in top_indices:
        retrieved_documents.append(documents[index])

    context = "\n\n".join(retrieved_documents)

    return context
```

```python
# ============================================
# SMARTAGRI AI - GEMINI FARMER ADVISORY
# ============================================

import time

# Gemini model
MODEL_NAME = "gemini-3.5-flash"


def ask_farmer(question):

    # Check empty question
    if not question.strip():
        return "Please enter a farmer question."

    # Retrieve relevant agricultural knowledge
    context = retrieve_context(question)

    # Prompt for Gemini
    prompt = f"""
You are SmartAgri AI, a helpful agricultural advisory assistant.

Answer the farmer's question using the agricultural knowledge
provided below.

IMPORTANT RULES:
1. Give a simple and clear answer.
2. Use the provided agricultural knowledge as the main source.
3. Do not invent unsupported agricultural facts.
4. If the information is not available, clearly say:
   "The available agricultural manual does not contain enough information."
5. Give practical advice in simple language.
6. Do not give dangerous or highly specific chemical recommendations.

AGRICULTURAL KNOWLEDGE:
{context}

FARMER QUESTION:
{question}

Give a short, clear and practical answer suitable for a farmer.
"""

    # Try 3 times if Gemini server is temporarily busy
    for attempt in range(3):

        try:

            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt
            )

            return response.text

        except Exception as e:

            error_message = str(e)

            # Temporary Gemini server error
            if "503" in error_message or "UNAVAILABLE" in error_message:

                if attempt < 2:

                    print(
                        "Gemini server is temporarily busy."
                    )

                    print(
                        f"Retrying... Attempt {attempt + 2}/3"
                    )

                    time.sleep(5)

                else:

                    return """
Gemini server is temporarily busy.

Please wait for a few seconds and run the question again.
Your dataset and project code are not affected.
"""

            else:

                return "Gemini Error: " + error_message


# ============================================
# TEST THE FARMER ADVISORY
# ============================================

question = "What nutrients are important for maize?"

answer = ask_farmer(question)

print("Farmer Question:")
print(question)

print("\nAI Advisory:")
print(answer)
```

```python
# ============================================================
# SMARTAGRI AI
# Crop Yield Forecasting + LLM Farmer Advisory Agent
# ============================================================

# ============================================================
# CELL 1 - INSTALL LIBRARIES
# ============================================================

!pip install -q google-genai gradio scikit-learn pandas numpy matplotlib seaborn


# ============================================================
# CELL 2 - IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import time

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("All libraries loaded successfully!")


# ============================================================
# CELL 3 - LOAD DATASET
# ============================================================

df = pd.read_csv("smartagri_crop_yield_dataset.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())


# ============================================================
# CELL 4 - DATA INSPECTION
# ============================================================

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Statistics:")
display(df.describe())


# ============================================================
# CELL 5 - EDA: CROP DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

df["Crop"].value_counts().plot(kind="bar")

plt.title("Crop Distribution")
plt.xlabel("Crop")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# CELL 6 - EDA: SOIL DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

df["Soil_Type"].value_counts().plot(kind="bar")

plt.title("Soil Type Distribution")
plt.xlabel("Soil Type")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# CELL 7 - AVERAGE YIELD BY CROP
# ============================================================

plt.figure(figsize=(8, 5))

df.groupby("Crop")["Yield"].mean().plot(kind="bar")

plt.title("Average Crop Yield")
plt.xlabel("Crop")
plt.ylabel("Average Yield")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# CELL 8 - RAINFALL VS YIELD
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(df["Rainfall"], df["Yield"])

plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall")
plt.ylabel("Yield")

plt.show()


# ============================================================
# CELL 9 - TEMPERATURE VS YIELD
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(df["Temperature"], df["Yield"])

plt.title("Temperature vs Crop Yield")
plt.xlabel("Temperature")
plt.ylabel("Yield")

plt.show()


# ============================================================
# CELL 10 - CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(10, 7))

numeric_data = df.select_dtypes(include=["number"])

correlation = numeric_data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Soil, Climate and Crop Yield")

plt.show()


# ============================================================
# CELL 11 - PREPARE ML DATA
# ============================================================

X = df.drop("Yield", axis=1)
y = df["Yield"]

print("Input Features:")
print(X.columns.tolist())

print("\nTarget:")
print("Yield")


# ============================================================
# CELL 12 - ENCODE CATEGORICAL FEATURES
# ============================================================

X = pd.get_dummies(
    X,
    columns=["Crop", "Soil_Type"]
)

print("Encoded dataset:")
display(X.head())

print("Number of features:", X.shape[1])


# ============================================================
# CELL 13 - TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))


# ============================================================
# CELL 14 - RANDOM FOREST MODEL
# ============================================================

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Random Forest model trained successfully!")


# ============================================================
# CELL 15 - MODEL PREDICTION
# ============================================================

y_pred = model.predict(X_test)

print("First 10 Predictions:")
print(y_pred[:10])


# ============================================================
# CELL 16 - MODEL EVALUATION
# ============================================================

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("\nMODEL PERFORMANCE")
print("-------------------------")
print("MAE  :", round(mae, 2))
print("MSE  :", round(mse, 2))
print("RMSE :", round(rmse, 2))
print("R2   :", round(r2, 2))


# ============================================================
# CELL 17 - ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Yield")
plt.ylabel("Predicted Yield")

plt.title("Actual vs Predicted Crop Yield")

plt.show()


# ============================================================
# CELL 18 - FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

display(feature_importance.head(10))


# ============================================================
# CELL 19 - FEATURE IMPORTANCE GRAPH
# ============================================================

top_features = feature_importance.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"][::-1],
    top_features["Importance"][::-1]
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.title("Top Factors Affecting Crop Yield")

plt.show()


# ============================================================
# CELL 20 - SAMPLE FARMER INPUT
# ============================================================

farmer_data = pd.DataFrame({
    "Crop": ["Rice"],
    "Soil_Type": ["Alluvial"],
    "Soil_pH": [6.5],
    "Nitrogen": [80],
    "Phosphorus": [40],
    "Potassium": [45],
    "Temperature": [28],
    "Rainfall": [1200],
    "Humidity": [80]
})

display(farmer_data)


# ============================================================
# CELL 21 - FARMER YIELD PREDICTION
# ============================================================

farmer_encoded = pd.get_dummies(
    farmer_data,
    columns=["Crop", "Soil_Type"]
)

farmer_encoded = farmer_encoded.reindex(
    columns=X.columns,
    fill_value=0
)

predicted_yield = model.predict(farmer_encoded)[0]

print(
    "Predicted Crop Yield:",
    round(predicted_yield, 2)
)


# ============================================================
# CELL 22 - AGRICULTURAL MANUAL / KNOWLEDGE BASE
# ============================================================

agriculture_knowledge = """

RICE

Rice requires suitable soil moisture and adequate nutrients.
Nitrogen is important for vegetative growth.
Phosphorus supports root development.
Potassium helps improve plant strength and stress tolerance.
Rice generally performs well in warm conditions with sufficient water.
Farmers should monitor irrigation, weeds, pests and diseases regularly.


WHEAT

Wheat grows well in suitable fertile soil and moderate temperatures.
Nitrogen supports leaf and plant growth.
Phosphorus supports root development.
Potassium contributes to overall plant health.
Proper irrigation is important during critical growth stages.
Farmers should monitor weeds and common diseases.


MAIZE

Maize requires adequate nitrogen for healthy vegetative growth.
Phosphorus supports root development and early plant growth.
Potassium supports plant strength and stress tolerance.
Suitable temperature and adequate rainfall are important.
Farmers should monitor soil moisture, weeds and pests.


COTTON

Cotton requires suitable soil conditions and adequate nutrients.
Nitrogen supports vegetative growth.
Phosphorus helps root development.
Potassium supports plant strength and boll development.
Farmers should monitor water availability and insect pests.


GROUNDNUT

Groundnut grows well in suitable well-drained soil.
Phosphorus supports root development and crop establishment.
Potassium contributes to plant health.
Calcium availability is important for pod development.
Farmers should avoid excessive waterlogging and monitor pests.


SUGARCANE

Sugarcane requires adequate water and nutrients.
Nitrogen supports vegetative growth.
Phosphorus supports root development.
Potassium contributes to plant strength and quality.
Regular irrigation and weed management are important.


MILLETS

Millets are generally adaptable crops and can grow under relatively dry conditions.
Balanced nutrients support healthy growth.
Farmers should monitor soil moisture, weeds and pests.
Suitable soil and climate conditions improve crop performance.


GENERAL CROP MANAGEMENT

Farmers should test soil before applying fertilizers.
Fertilizer application should be based on soil condition and crop requirement.
Irrigation should be managed according to crop stage and weather conditions.
Farmers should regularly monitor pests and diseases.
Weed management is important during crop growth.
Weather conditions such as temperature and rainfall can affect crop yield.
"""


# ============================================================
# CELL 23 - CREATE KNOWLEDGE DOCUMENTS
# ============================================================

documents = [
    section.strip()
    for section in agriculture_knowledge.split("\n\n")
    if section.strip()
]

print("Knowledge sections:", len(documents))


# ============================================================
# CELL 24 - TF-IDF KNOWLEDGE RETRIEVAL
# ============================================================

vectorizer = TfidfVectorizer(
    stop_words="english"
)

document_vectors = vectorizer.fit_transform(documents)

print("Agricultural manual indexed successfully!")


# ============================================================
# CELL 25 - RETRIEVAL FUNCTION
# ============================================================

def retrieve_context(question, top_k=2):

    question_vector = vectorizer.transform([question])

    similarity_scores = cosine_similarity(
        question_vector,
        document_vectors
    ).flatten()

    top_indices = similarity_scores.argsort()[::-1][:top_k]

    retrieved_documents = []

    for index in top_indices:
        retrieved_documents.append(
            documents[index]
        )

    context = "\n\n".join(
        retrieved_documents
    )

    return context


# ============================================================
# CELL 26 - TEST RETRIEVAL
# ============================================================

question = "What nutrients are important for maize?"

context = retrieve_context(question)

print("Retrieved Manual Information:")
print("-----------------------------")
print(context)


# ============================================================
# CELL 27 - GEMINI API CONNECTION
# ============================================================

from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

if not api_key:

    raise ValueError(
        "GEMINI_API_KEY not found. "
        "Please add GEMINI_API_KEY in Colab Secrets."
    )

client = genai.Client(
    api_key=api_key
)

print("Gemini connected successfully!")


# ============================================================
# CELL 28 - LLM FARMER ADVISORY FUNCTION
# ============================================================

MODEL_NAME = "gemini-3.5-flash"


def ask_farmer(question):

    if not question.strip():

        return "Please enter a farmer question."


    # Retrieve relevant manual information

    context = retrieve_context(question)


    # Create prompt

    prompt = f"""
You are SmartAgri AI,
an agricultural advisory assistant.

Answer the farmer's question using the
agricultural manual information provided below.

IMPORTANT RULES:

1. Give a simple and clear answer.
2. Use the provided manual as the main source.
3. Do not invent unsupported agricultural facts.
4. If the information is unavailable, say:
   "The available agricultural manual does not
   contain enough information."
5. Give practical advice in simple language.
6. Do not provide dangerous or highly specific
   chemical recommendations.

AGRICULTURAL MANUAL:

{context}


FARMER QUESTION:

{question}


Give a short and clear answer suitable
for a farmer.
"""


    # Retry if Gemini temporarily unavailable

    for attempt in range(3):

        try:

            response = client.models.generate_content(

                model=MODEL_NAME,

                contents=prompt
            )

            return response.text


        except Exception as e:

            error_message = str(e)


            if (
                "503" in error_message
                or
                "UNAVAILABLE" in error_message
            ):

                if attempt < 2:

                    print(
                        "Gemini server temporarily busy."
                    )

                    print(
                        f"Retrying... "
                        f"{attempt + 2}/3"
                    )

                    time.sleep(5)

                else:

                    return (
                        "Gemini server is temporarily busy. "
                        "Please try again after a few seconds."
                    )


            else:

                return (
                    "Gemini Error: "
                    + error_message
                )


# ============================================================
# CELL 29 - TEST LLM
# ============================================================

question = "What nutrients are important for maize?"

answer = ask_farmer(question)

print("Farmer Question:")
print(question)

print("\nAI Advisory:")
print(answer)


# ============================================================
# CELL 30 - VISUAL/TEXT FARMER ADVISORY AGENT
# ============================================================

import gradio as gr


def farmer_advisor(question):

    if not question.strip():

        return "Please enter a farmer question."

    try:

        return ask_farmer(question)

    except Exception as e:

        return "Error: " + str(e)


with gr.Blocks() as demo:

    gr.Markdown(
        """
        # 🌾 SmartAgri AI

        ## Crop Yield Forecasting & Farmer Advisory System

        Ask questions about crop management
        using natural language.
        """
    )


    gr.Markdown(
        """
        ### Example Questions

        🌱 What nutrients are important for maize?

        🌾 How should rice be managed?

        🌿 What nutrients are needed for wheat?

        🥜 How can I manage groundnut cultivation?

        🌦️ What factors affect crop yield?
        """
    )


    question_box = gr.Textbox(

        label="👨‍🌾 Farmer Question",

        placeholder=
        "Example: What nutrients are important for maize?",

        lines=3
    )


    ask_button = gr.Button(
        "🌱 Ask SmartAgri AI"
    )


    answer_box = gr.Markdown(
        label="🤖 AI Advisory"
    )


    ask_button.click(

        fn=farmer_advisor,

        inputs=question_box,

        outputs=answer_box
    )


demo.launch(
    share=True
)
```
