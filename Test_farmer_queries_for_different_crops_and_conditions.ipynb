```python
# ============================================================
# TEST FARMER QUERIES FOR DIFFERENT CROPS AND CONDITIONS
# ============================================================

import pandas as pd

# Load your actual dataset
df = pd.read_csv("/content/smartagri_crop_yield_dataset.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nAvailable Crops:")
print(df["Crop"].unique())


# ------------------------------------------------------------
# Create crop-based knowledge from YOUR dataset
# ------------------------------------------------------------

crop_info = df.groupby("Crop").agg({
    "Soil_Type": lambda x: x.mode()[0],
    "Soil_pH": "mean",
    "Nitrogen": "mean",
    "Phosphorus": "mean",
    "Potassium": "mean",
    "Temperature": "mean",
    "Rainfall": "mean",
    "Humidity": "mean",
    "Yield": "mean"
}).reset_index()


# ------------------------------------------------------------
# Farmer Query Function
# ------------------------------------------------------------

def farmer_advisory(question):

    question_lower = question.lower()

    # Find crop
    selected_crop = None

    for crop in df["Crop"].unique():

        if crop.lower() in question_lower:
            selected_crop = crop
            break

    # Crop not found
    if selected_crop is None:
        return "Crop information is not available in the dataset."

    # Get crop information
    info = crop_info[
        crop_info["Crop"] == selected_crop
    ].iloc[0]

    # Soil query
    if "soil" in question_lower:

        return (
            f"{selected_crop} generally uses "
            f"{info['Soil_Type']} soil in this dataset. "
            f"Average soil pH is {info['Soil_pH']:.2f}."
        )

    # Rainfall query
    elif "rainfall" in question_lower or "rain" in question_lower:

        return (
            f"The average rainfall recorded for "
            f"{selected_crop} is {info['Rainfall']:.2f}."
        )

    # Temperature query
    elif "temperature" in question_lower or "heat" in question_lower:

        return (
            f"The average temperature recorded for "
            f"{selected_crop} is {info['Temperature']:.2f}."
        )

    # Nitrogen query
    elif "nitrogen" in question_lower:

        return (
            f"The average nitrogen value for "
            f"{selected_crop} is {info['Nitrogen']:.2f}."
        )

    # Phosphorus query
    elif "phosphorus" in question_lower:

        return (
            f"The average phosphorus value for "
            f"{selected_crop} is {info['Phosphorus']:.2f}."
        )

    # Potassium query
    elif "potassium" in question_lower:

        return (
            f"The average potassium value for "
            f"{selected_crop} is {info['Potassium']:.2f}."
        )

    # Humidity query
    elif "humidity" in question_lower:

        return (
            f"The average humidity recorded for "
            f"{selected_crop} is {info['Humidity']:.2f}."
        )

    # Yield query
    elif "yield" in question_lower:

        return (
            f"The average yield recorded for "
            f"{selected_crop} is {info['Yield']:.2f}."
        )

    # General crop query
    else:

        return (
            f"{selected_crop} information: "
            f"Soil = {info['Soil_Type']}, "
            f"pH = {info['Soil_pH']:.2f}, "
            f"Rainfall = {info['Rainfall']:.2f}, "
            f"Temperature = {info['Temperature']:.2f}, "
            f"Humidity = {info['Humidity']:.2f}."
        )


# ============================================================
# TEST DIFFERENT FARMER QUERIES
# ============================================================

queries = [

    "What soil is suitable for rice?",
    "What is the rainfall for rice?",

    "What is the soil condition for wheat?",
    "What temperature is suitable for wheat?",

    "What is the nitrogen level for maize?",
    "What is the rainfall for maize?",

    "What is the potassium level for cotton?",
    "What is the humidity for cotton?",

    "What is the soil pH for tomato?",
    "What is the yield of tomato?",

    "What is the phosphorus level for groundnut?",
    "What is the rainfall for groundnut?"
]


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("\n================================================")
print("        FARMER QUERY TESTING")
print("================================================")

successful = 0

for i, question in enumerate(queries, 1):

    answer = farmer_advisory(question)

    print(f"\nQuery {i}:")
    print("Farmer:", question)

    print("Assistant:", answer)

    if "not available" not in answer:
        successful += 1


# ============================================================
# TEST UNSUPPORTED CROP
# ============================================================

print("\n================================================")
print("        UNSUPPORTED QUERY TEST")
print("================================================")

unsupported_query = "What soil is suitable for banana?"

print("Farmer:", unsupported_query)

print(
    "Assistant:",
    farmer_advisory(unsupported_query)
)


# ============================================================
# TESTING SUMMARY
# ============================================================

total_queries = len(queries)

success_percentage = (
    successful / total_queries
) * 100

print("\n================================================")
print("              TEST SUMMARY")
print("================================================")

print("Total queries tested:", total_queries)
print("Successful queries:", successful)
print("Success percentage:", round(success_percentage, 2), "%")

print("\nFarmer query testing completed successfully!")
```
